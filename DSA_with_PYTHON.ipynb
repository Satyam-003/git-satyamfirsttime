```python
class LinkedList:
  def __init__(self):
    self.head = None
    self.n = 0 #we are doing this we want to keep track that how many nodes are there

  def __len__(self):
    return self.n

  def insert_head(self,value):
    # Creating a new node
    New_Node = Node(value)

    # Making the connection
    New_Node.next = self.head

    # Re-assigning the head to new node
    self.head = New_Node

    self.n += 1

# Traversing through LL
  def __str__(self):
    curr = self.head
    result = ''
    while curr != None:
      result = result + str(curr.data) + '->'
      curr = curr.next

    return(result[:-2])


  def append(self, value):

    new_node = Node(value)

    if self.head == None:
      self.head = new_node
      self.n += 1
      return

    curr = self.head
    while curr.next != None:
      curr = curr.next

    curr.next = new_node
    self.n += 1

L = LinkedList() # creating the object L

L.insert_head(1)
L.insert_head(2)
L.insert_head(3)
L.insert_head(4)
L.append(5)
print('The length of LL is :',len(L))
print(L)
```
